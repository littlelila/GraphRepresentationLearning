```python
from pathlib import Path

import pandas as pd
import obonet
import networkx as nx

PROJECT_ROOT = Path("..")

HPO_PATH = PROJECT_ROOT / "raw" / "hp.obo"
HPOA_PATH = PROJECT_ROOT / "raw" / "phenotype.hpoa"
```

```python
hpo = obonet.read_obo(HPO_PATH)

print("HPO nodes:", len(hpo.nodes))
print("HPO edges:", len(hpo.edges))
```

```python
for u, v, data in list(hpo.edges(data=True))[:10]:
    print(u, "->", v, data)
```

```python
hpo_labels = {
    node_id: data.get("name", node_id)
    for node_id, data in hpo.nodes(data=True)
}

for child, parent, data in list(hpo.edges(data=True))[:10]:
    print(
        child,
        hpo_labels.get(child, child),
        "->",
        parent,
        hpo_labels.get(parent, parent)
    )
```

```python
hpoa = pd.read_csv(
    HPOA_PATH,
    sep="\t",
    comment="#",
    dtype=str
)

print("Rows:", len(hpoa))
print("Columns:")
print(hpoa.columns.tolist())

hpoa.head()
```

```python
hpoa = hpoa[hpoa["qualifier"].fillna("") != "NOT"]

if "aspect" in hpoa.columns:
    hpoa = hpoa[hpoa["aspect"] == "P"]

print("Rows after filtering:", len(hpoa))
print("Diseases:", hpoa["database_id"].nunique())
print("HPO terms:", hpoa["hpo_id"].nunique())
```
